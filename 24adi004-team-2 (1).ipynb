```python
# LIBRARIES USED

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# DATA LOADING

df=pd.read_csv("/kaggle/input/life-expectancy-who/Life Expectancy Data.csv")
df.columns = df.columns.str.strip()

print("Dataset Shape:", df.shape)

print("First 5 Rows")
print(df.head())

print("Last 5 Rows")
print(df.tail())

print("Columns in Dataset")
print(df.columns)
```

```python
# WEEK 2 - KNOW YOUR DATA

print("Data Types")
print(df.dtypes)

print("Statistical Summary")
display(df.describe())

print("Missing Values")
print(df.isnull().sum().sort_values(ascending=False))

print("Unique Information")
print("Total Countries:",df["Country"].nunique())
print("Status Types:",df["Status"].unique())
print("Year Range:",df["Year"].min(),"-",df["Year"].max())

print("First Impression Report")
print("Rows:",df.shape[0])
print("Columns:",df.shape[1])
print("Missing Values Present:",df.isnull().values.any())
print("Duplicate Rows:",df.duplicated().sum())
```

```python
# WEEK 3 - CLEANING

print("Dataset Shape Before Cleaning:",df.shape)

missing_before=df.isnull().sum()
print("Missing Values Before Cleaning")
print(missing_before)

missing_nonzero=missing_before[missing_before>0]

if len(missing_nonzero)>0:
    plt.figure(figsize=(12,5))
    missing_nonzero.plot(kind="bar",color="orange")
    plt.title("Missing Values Before Imputation")
    plt.ylabel("Count")
    plt.show()

num_cols=df.select_dtypes(include=np.number).columns
cat_cols=df.select_dtypes(include="object").columns

print("Numerical Columns")
print(list(num_cols))

print("Categorical Columns")
print(list(cat_cols))

print("Applying Imputation Strategy")

for col in num_cols:
    if df[col].skew()>1:
        df[col]=df[col].fillna(df[col].median())
    else:
        df[col]=df[col].fillna(df[col].mean())

for col in cat_cols:
    df[col]=df[col].fillna(df[col].mode()[0])

print("Missing Values After Cleaning")
print(df.isnull().sum())
```

```python
# Outlier Detection using IQR Method

important_cols=["Life expectancy","Adult Mortality","Alcohol"]
df_cleaned=df.copy()

for col in important_cols:
    Q1=df_cleaned[col].quantile(0.25)
    Q3=df_cleaned[col].quantile(0.75)
    IQR=Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    df_cleaned=df_cleaned[(df_cleaned[col]>=lower)&(df_cleaned[col]<=upper)]

print("Dataset Shape Comparison")
print("Before Removing Outliers:",df.shape)
print("After Removing Outliers:",df_cleaned.shape)
```

```python
# COMPARISION BETWEEN THE DATA SET PRE OUTLIER REMOVAL AND POST OUTLIER REMOVAL

plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
sns.boxplot(x=df["Life expectancy"],color="skyblue")
plt.title("Before Outlier Removal")

plt.subplot(1,2,2)
sns.boxplot(x=df_cleaned["Life expectancy"],color="lightgreen")
plt.title("After Outlier Removal")

plt.tight_layout()
plt.show()

print("Final Cleaned Dataset Summary")
display(df_cleaned.describe())
```

```python
# WEEK 4 - EDA

df=df_cleaned.copy()

sns.set_style("whitegrid")
sns.set_palette("Set2")
```

```python
# UNIVARIATE ANALYSIS 

print("UNIVARIATE ANALYSIS")
plt.figure(figsize=(8,5))
sns.histplot(df["Life expectancy"],bins=25,kde=True,color="#2E86C1")
plt.title("Distribution of Life Expectancy")
plt.xlabel("Life Expectancy")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x=df["Life expectancy"],color="lightblue")
plt.title("Boxplot of Life Expectancy")
plt.show()
```

```python
# BIVARIATE ANALYSIS

print("BIVARIATE ANALYSIS")

# GDP Group vs Life Expectancy
df["GDP_group"]=pd.qcut(df["GDP"],4,labels=["Low","Lower-Mid","Upper-Mid","High"])
gdp_avg=df.groupby("GDP_group",observed=False)["Life expectancy"].mean()
plt.figure(figsize=(7,5))
plt.plot(gdp_avg.index,gdp_avg.values,marker="o",color="purple")
plt.title("Average Life Expectancy by GDP Group")
plt.xlabel("GDP Group")
plt.ylabel("Average Life Expectancy")
plt.show()

# Schooling vs Life Expectancy
plt.figure(figsize=(8,6))
sns.scatterplot(x="Schooling",y="Life expectancy",data=df,hue="Status",alpha=0.6)
plt.title("Schooling vs Life Expectancy")
plt.show()

# Life Expectancy by Status
status_avg=df.groupby("Status")["Life expectancy"].mean()
plt.figure(figsize=(6,5))
plt.bar(status_avg.index,status_avg.values,color=["orange","green"])
plt.title("Average Life Expectancy by Country Status")
plt.ylabel("Life Expectancy")
plt.show()

# Adult Mortality vs Life Expectancy 
plt.figure(figsize=(8,6))
sns.scatterplot(x="Adult Mortality",y="Life expectancy",data=df,hue="Status",alpha=0.6)
plt.title("Adult Mortality vs Life Expectancy")
plt.show()

# HIV/AIDS vs Life Expectancy
plt.figure(figsize=(8,6))
sns.scatterplot(x="HIV/AIDS",y="Life expectancy",data=df,hue="Status",alpha=0.6)
plt.title("HIV/AIDS vs Life Expectancy")
plt.show()
```

```python
# CORRELATION ANALYSIS

plt.figure(figsize=(12,8))
corr=df.corr(numeric_only=True)
sns.heatmap(corr,cmap="coolwarm",linewidths=0.5)
plt.title("Correlation Matrix")
plt.show()
```

```python
# YEARWISE TREND 

year_avg=df.groupby("Year")["Life expectancy"].mean()
plt.figure(figsize=(8,5))
plt.plot(year_avg.index,year_avg.values,marker="o",color="black")
plt.title("Average Life Expectancy Over Years")
plt.xlabel("Year")
plt.ylabel("Average Life Expectancy")
plt.grid(True)
plt.show()
```
